```python
df=pd.read_csv('Clean_Dataset.csv')
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
print(df.columns)
```

```python
df = df.fillna(0)
df = df.dropna()
print(df.shape)
```

```python
#convert departure time categories into numerical values
departure_time_map = {
    "Early_Morning": 0,
    "Morning": 1,
    "Afternoon": 2,
    "Evening": 3,
    "Night": 4,
    "Late_Night": 5
}
df["departure_time"] = df["departure_time"].map(departure_time_map)
```

```python
#convert arrival time categories into numerical values
arrival_time_map = {
    "Early_Morning": 0,
    "Morning": 1,
    "Afternoon": 2,
    "Evening": 3,
    "Night": 4,
    "Late_Night": 5
}
df["arrival_time"] = df["arrival_time"].map(arrival_time_map)
```

```python
stops_map = {
    "non-stop": 0,
    "1 stop": 1,
    "2 stops": 2,
    "3 stops": 3,
    "4 stops": 3
}
df["stops"] = df["stops"].map(stops_map)
```

```python
df = df.fillna(0)
df = df.dropna()
print(df.shape)
df.head()
```

```python
#scatter plot
plt.scatter(df['duration'],df['price'], c='red', s=2)
plt.xlabel("Duration of Flight")
plt.ylabel("Price of Flight")
```

```python
k=2
X=np.array(list(zip(df['duration'], df['price'])))
model=KMeans(n_clusters=k)
kmeans=model.fit(X)
labels=model.predict(X)
centroids=model.cluster_centers_

print(labels)
print(centroids)
```

```python
#Plot the data and centroids on a scatter plot
#map the labels to colors

c=['m','g','b','r','y','c']
colors=[c[i] for i in labels]

plt.scatter(df['duration'], df['price'], c=colors, s=2)
plt.scatter(centroids[:,0], centroids[:,1], marker='*', s=100, c='black')
```

```python
#Using the model that you have just trained,
#you can use it to predict what cluster for any data pt

cluster=model.predict([[35,123]])[0]
print(c[cluster])
```

```python
#To determine the silhouette coefficient/score

from sklearn.metrics import silhouette_score

model=KMeans(n_clusters=2)
model.fit(X)
label=model.predict(X)

print(f'silhouette Score (n=2): {silhouette_score(X,label)}')


model=KMeans(n_clusters=3)
model.fit(X)
label=model.predict(X)

print(f'silhouette Score (n=3): {silhouette_score(X,label)}')


model=KMeans(n_clusters=4)
model.fit(X)
label=model.predict(X)

print(f'silhouette Score (n=4): {silhouette_score(X,label)}')
```

```python
k=4
X=np.array(list(zip(df['duration'],df['price'])))

kmeans=KMeans(n_clusters=k)
kmeans=kmeans.fit(X)
labels=kmeans.predict(X)
centroids=kmeans.cluster_centers_

#map the labels to colors
c=['b','r','y','g','c','m']
colors=[c[i] for i in labels]

plt.scatter(df['duration'],df['price'],c=colors,s=2)
plt.scatter(centroids[:,0],centroids[:,1],marker='*',s=100,c='black')

print(centroids)
```

# This analysis works with a cleaned airline flight dataset that includes details such as flight duration, ticket price, departure and arrival times, and number of stops. To make the data usable for analysis, missing values were handled and descriptive categories (like “Morning” or “1 stop”) were converted into simple numbers so the computer could process them. For the clustering step, we focused on two business-friendly factors: **flight duration** and **ticket price**, and visualized their relationship with a scatter plot. We then applied **K-Means clustering**, which groups similar flights together based on these two characteristics. Different values of **k** (the number of groups) were tested using a silhouette score to measure how well the flights were separated, and **k = 4** was selected as it provided a clear and meaningful structure. The final result divides all flights into four distinct groups, each represented by a centroid that reflects a typical price–duration combination. In plain terms, this means the flights can be categorized into four pricing patterns (for example, short and cheap, long and expensive, and two groups in between), giving the business an easy way to understand and compare flight options without needing any background in machine learning.

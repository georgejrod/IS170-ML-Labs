```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
lr_model = LinearRegression()
lr_model.fit(x_train, y_train)

def get_prediction():
    cols = ['Model','Mileage','Year','EngineV','Brand','Body','Engine Type']
    while True:
        try:
            model_code = int(input("Enter Model code: "))
            mileage = float(input("Enter Mileage: "))
            year = int(input("Enter Year: "))
            enginev = float(input("Enter EngineV: "))
            brand_code = int(input("Enter Brand code: "))
            body_code = int(input("Enter Body code: "))
            engine_type_code = int(input("Enter Engine Type code: "))

            input_df = pd.DataFrame(
                [[model_code, mileage, year, enginev, brand_code, body_code, engine_type_code]],
                columns=cols
            )

            predicted_price = lr_model.predict(input_df)[0]
            print(f"Predicted price: ${predicted_price:.2f}")
            break


        except ValueError:
            print("Invalid input. Please enter numeric values.")
        except Exception as e:
            print(f"An error occurred: {e}")
            break

while True:
    get_prediction()
    another = input("Do you want to predict another price? (yes/no): ")
    if another.lower() != "yes":
        break
```

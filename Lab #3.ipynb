```python
import pandas as pd
import numpy as np
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# Read the Kaggle dataset (it already has headers)
df = pd.read_csv('products.csv')
print(df.shape)
```

```python
# Confirm columns / shape
print(df.shape)
print(df.columns)
df.head()
```

```python
# Build transactions from the 'Products' column
# Items are comma-separated in the file (often ", ").
# Also handle possible missing values safely.

transactions = (
    df["Products"]
    .fillna("")
    .astype(str)
    .apply(lambda s: [item.strip() for item in s.split(",") if item.strip() != ""])
    .tolist()
)

transactions[:3]
```

```python
# One-hot encode transactions
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
df_hot = pd.DataFrame(te_data, columns=te.columns_).astype(int)

df_hot.head()
```

```python
# Frequent itemsets (SUPPORT)
# Use a smaller support first; 0.2 is usually too high for 30k retail transactions.
freq_itemsets = apriori(df_hot, min_support=0.01, use_colnames=True)
freq_itemsets.sort_values("support", ascending=False).head(10)
```

```python
# Association rules: CONFIDENCE
rules_conf = association_rules(freq_itemsets, metric="confidence", min_threshold=0.6)
rules_conf.sort_values("confidence", ascending=False).head(10)
```

```python
# Association rules: LIFT
rules_lift = association_rules(freq_itemsets, metric="lift", min_threshold=1.0)
rules_lift.sort_values("lift", ascending=False).head(10)
```

```python
# Association rules: LIFT
rules_lift = association_rules(freq_itemsets, metric="lift", min_threshold=1.0)
rules_lift.sort_values("lift", ascending=False).head(10)
```

```python
# Association rules: CONVICTION
rules_conv = association_rules(freq_itemsets, metric="conviction", min_threshold=1.0)
rules_conv.sort_values("conviction", ascending=False).head(10)
```

# **Executive Summary of Market Basket Analysis Results**
We analyzed customer transaction data to better understand which products are frequently purchased together. Using an approach called market basket analysis, we identified patterns in buying behavior that can help guide promotions, product placement, bundling strategies, and cross-selling opportunities.

At a high level, the analysis tells us which product combinations appear together more often than we would expect by chance. For example, if customers who buy Product A also frequently buy Product B, that creates an opportunity to promote those items together, place them near each other in-store, or suggest one when the other is added to an online cart.

# **Main Measures**

1. Support (Popularity of a Combination)
Support tells us how common a product combination is across all transactions.

For example, if “Bread & Milk” has a support of 5%, that means 5% of all transactions included both items together. For management, Support shows us how big the opportunity is. Higher support means the combination affects more customers.

---

2. Confidence (Likelihood of Buying Together)
Confidence answers the question:
“If a customer buys Product A, how likely are they to also buy Product B?”

For example, if confidence is 70%, that means 7 out of 10 customers who bought Product A also bought Product B. For managementm Confidence tells us how strong the cross-sell relationship is.

---

3. Lift (Strength Beyond Coincidence)
Lift tells us whether two products are purchased together more often than we would expect by random chance.

Lift = 1 → No real relationship

Lift > 1 → Positive relationship (they truly go together)

Lift < 1 → Negative relationship

For management, Lift tells us whether the pairing is meaningful or just random overlap.

---

4. Conviction (Reliability of the Rule)
Conviction measures how dependable the relationship is. Higher conviction values indicate stronger and more reliable associations.

For management, Conviction helps confirm that the pattern is stable and not accidental.
